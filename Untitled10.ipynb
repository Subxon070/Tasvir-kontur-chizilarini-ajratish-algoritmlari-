```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

def process_image(image_name, title_prefix):
    print(f"\n--- {title_prefix} uchun ishlov berish boshlandi: {image_name} ---")
    img = cv2.imread(image_name)
    if img is None:
        print("Xatolik: Rasm topilmadi!")
        return

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 1. SOBEL ALGORITMI
    sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
    abs_x = cv2.convertScaleAbs(sobel_x)
    abs_y = cv2.convertScaleAbs(sobel_y)
    sobel_combined = cv2.addWeighted(abs_x, 0.5, abs_y, 0.5, 0)
    res_sobel = 255 - sobel_combined

    # 2. UOLLS ALGORITMI
    kernel_wx = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]], dtype=np.float32)
    kernel_wy = np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]], dtype=np.float32)
    gx_w = cv2.filter2D(gray, cv2.CV_32F, kernel_wx)
    gy_w = cv2.filter2D(gray, cv2.CV_32F, kernel_wy)
    mag_w = cv2.sqrt(gx_w*gx_w + gy_w*gy_w)
    mag_w = cv2.convertScaleAbs(mag_w)
    res_uolls = 255 - mag_w

    # 3. CANNY ALGORITMI
    blur = cv2.GaussianBlur(gray, (5, 5), 1.4)
    edges = cv2.Canny(blur, 100, 200)
    res_canny = 255 - edges

    # 4. ROBERTS ALGORITMI
    kernel_roberts_x = np.array([[1, 0], [0, -1]], dtype=np.float32) # yoki mos ravishda
    kernel_roberts_y = np.array([[0, 1], [-1, 0]], dtype=np.float32)
    gx_r = cv2.filter2D(gray, cv2.CV_32F, np.array([[1, 0], [0, 1]], dtype=np.float32)) # soddalashtirilgan
    gy_r = cv2.filter2D(gray, cv2.CV_32F, np.array([[0, 1], [-1, 0]], dtype=np.float32))
    # Asl Roberts matritsasi bo'yicha:
    k_rx = np.array([[1, 0], [0, -1]], dtype=np.float32)
    k_ry = np.array([[0, 1], [-1, 0]], dtype=np.float32)
    rx = cv2.filter2D(gray, cv2.CV_32F, k_rx)
    ry = cv2.filter2D(gray, cv2.CV_32F, k_ry)
    mag_r = cv2.sqrt(rx*rx + ry*ry)
    mag_r = cv2.convertScaleAbs(mag_r)
    res_roberts = 255 - mag_r

    # Natijalarni ko'rsatish
    print(f"--- {title_prefix}: Sobel natijasi ---")
    cv2_imshow(res_sobel)

    print(f"--- {title_prefix}: Uolls natijasi ---")
    cv2_imshow(res_uolls)

    print(f"--- {title_prefix}: Canny natijasi ---")
    cv2_imshow(res_canny)

    print(f"--- {title_prefix}: Roberts natijasi ---")
    cv2_imshow(res_roberts)

# --- 1-QADAM: Yuz rasmini yuklash va tekshirish ---
print("Iltimos, YUZ rasmini yuklang:")
uploaded_face = files.upload()
face_name = list(uploaded_face.keys())[0]
process_image(face_name, "Yuz rasmi")

# --- 2-QADAM: Barmoq izi rasmini yuklash va tekshirish ---
print("\nIltimos, BARMOQ IZI rasmini yuklang:")
uploaded_finger = files.upload()
finger_name = list(uploaded_finger.keys())[0]
process_image(finger_name, "Barmoq izi rasmi")
```
